```python
# Read one NORMAL image
normal_folder = os.path.join(train_path, "NORMAL")
normal_image_name = os.listdir(normal_folder)[0]
normal_image = cv2.imread(os.path.join(normal_folder, normal_image_name))
normal_image = cv2.cvtColor(normal_image, cv2.COLOR_BGR2RGB)

# Read one PNEUMONIA image
pneumonia_folder = os.path.join(train_path, "PNEUMONIA")
pneumonia_image_name = os.listdir(pneumonia_folder)[0]
pneumonia_image = cv2.imread(os.path.join(pneumonia_folder, pneumonia_image_name))
pneumonia_image = cv2.cvtColor(pneumonia_image, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(normal_image)
plt.title("NORMAL")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(pneumonia_image)
plt.title("PNEUMONIA")
plt.axis("off")

plt.show()
```
